```python
import pandas as pd
import nltk
import string
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Descargar recursos de NLTK
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

# Leer archivo
df = pd.read_excel('/content/sample_data/SoloResumen.xlsx')

# Asegura nombres de columnas en minúscula por consistencia
df.columns = df.columns.str.lower()

# Unir columnas de texto relevantes
df['texto_completo'] = df[['abstract']].fillna('').agg(' '.join, axis=1)

# Inicializar lematizador y stopwords
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

# Stopwords personalizadas adicionales (basadas en tu análisis previo)
stop_words_extra = [
    'species', 'based', 'using', 'however', 'result', 'study', 'different', 'also', 'within', 'found', 'used',
    'table', 'many', 'still', 'among', 'amongst', 'across', 'although', 'show', 'test', 'term', 'type',
    'may', 'might', 'could', 'would', 'should', 'can', 'will', 'one', 'two', 'three', 'four', 'example',
    'well', 'much', 'high', 'low', 'value', 'number', 'set', 'data', 'figure', 'etc', 'first', 'second',
    'important', 'resulted', 'resulting', 'observed', 'described', 'determined', 'obtained', 'provide',
    'included', 'defined', 'considered', 'presented', 'reported', 'total', 'tree', 'year', 'time', 'fig', 'plot', 'host', 'year', 'large', 'new', 'part',
    'rate', 'type', 'mean', 'index', 'tree', 'size', 'part', 'use', 'due', 'site', 'area', 'type', 'common',
    'survey', 'local', 'year', 'method', 'plot', 'site',
]
stop_words.update(stop_words_extra)

# Función de preprocesamiento
def preprocess(text):
    text = text.lower()
    text = ''.join([c for c in text if c not in string.punctuation])
    tokens = nltk.word_tokenize(text)
    # Aplicar lematización y filtro de stopwords
    tokens = [lemmatizer.lemmatize(w) for w in tokens if w not in stop_words and w.isalpha() and len(w) > 2]
    return ' '.join(tokens)

# Aplicar preprocesamiento
df['texto_limpio'] = df['texto_completo'].apply(preprocess)

# Concatenar todo el texto procesado
all_text = ' '.join(df['texto_limpio'])

# Crear y mostrar la nube de palabras
wordcloud = WordCloud(width=900, height=500, background_color='white', colormap='viridis', random_state=42).generate(all_text)

plt.figure(figsize=(15, 7))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```

```python
def tokenize(text):
    tokens = word_tokenize(text)
    return [word for word in tokens if word not in stop_words and len(word) > 2]

# Vectorización
vectorizer = CountVectorizer(
    tokenizer=tokenize,
    max_df=0.85,
    min_df=2,
    max_features=1000
)
# Assuming 'texto_limpio' is the column you want to use
dtm = vectorizer.fit_transform(df['texto_limpio'])
```

```python
lda = LatentDirichletAllocation(
    n_components=5,
    random_state=42,
    learning_method='batch'
)
lda.fit(dtm)

# Función para mostrar tópicos
def print_topics(model, vectorizer, n_words=10):
    feature_names = vectorizer.get_feature_names_out()
    for topic_idx, topic in enumerate(model.components_):
        print(f"Tópico #{topic_idx + 1}:")
        print(" ".join([
            feature_names[i] for i in topic.argsort()[:-n_words - 1:-1]
        ]))
        print("\n")

# Mostrar tópicos
print_topics(lda, vectorizer)
```

```python
import itertools
from collections import Counter
import ast
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
from networkx.algorithms import community
import pandas as pd

# Función para limpiar las palabras clave
def clean_keywords(x):
    if isinstance(x, str):
        try:
            x = ast.literal_eval(x)
        except:
            x = x.split(",")
    return [w.strip("[]'\" ").lower() for w in x if w.strip()]

dfOP["keywords"] = dfOP["keywords"].fillna("").apply(clean_keywords)

# Contar frecuencia de cada palabra clave en todos los artículos
keyword_counts = Counter(word for keywords in dfOP["keywords"] for word in set(keywords))

# Definir umbral mínimo de aparición de una palabra en distintos artículos
min_keyword_occurrence = 2
filtered_keywords = {word for word, count in keyword_counts.items() if count >= min_keyword_occurrence}

# Generar pares de coocurrencia solo con palabras clave frecuentes
coocurrence_pairs = []
for keywords in dfOP["keywords"]:
    filtered = list(set(keywords) & filtered_keywords)
    coocurrence_pairs.extend(itertools.combinations(filtered, 2))

# Contar la frecuencia de coocurrencia
coocurrence_counts = Counter(coocurrence_pairs)

# Filtrar conexiones débiles
min_coocurrence = 1
filtered_edges = {pair: count for pair, count in coocurrence_counts.items() if count >= min_coocurrence}

# Crear grafo de coocurrencia
G = nx.Graph()
for (word1, word2), count in filtered_edges.items():
    G.add_edge(word1, word2, weight=count)

# Verificar si la red es lo suficientemente grande
if len(G.nodes) < 2:
    print("La red de coocurrencia tiene muy pocos nodos para ser graficada.")
else:
    # Calcular métricas
    communities = community.louvain_communities(G, seed=42)
    degree_centrality = nx.degree_centrality(G)
    betweenness_centrality = nx.betweenness_centrality(G)

    # Crear mapeo de nodos a comunidades
    node_to_community = {}
    for i, comm in enumerate(communities):
        for node in comm:
            node_to_community[node] = i

    # ==================== OPCIÓN 1: LAYOUT CIRCULAR MEJORADO ====================
    plt.figure(figsize=(10, 8))

    # Layout circular agrupado por comunidades
    pos = {}
    angle_step = 2 * np.pi / len(communities)

    for i, comm in enumerate(communities):
        # Posición central de cada comunidad
        center_angle = i * angle_step
        center_x = 4 * np.cos(center_angle)
        center_y = 4 * np.sin(center_angle)

        # Distribuir nodos de la comunidad en círculo pequeño
        comm_list = list(comm)
        if len(comm_list) == 1:
            pos[comm_list[0]] = (center_x, center_y)
        else:
            comm_angle_step = 2 * np.pi / len(comm_list)
            radius = min(2.0, len(comm_list) * 0.25 + 0.5)
            for j, node in enumerate(comm_list):
                angle = j * comm_angle_step
                pos[node] = (center_x + radius * np.cos(angle),
                           center_y + radius * np.sin(angle))

    # Configurar colores para comunidades
    colors = ['#D95F43', '#569981', '#43D9A3', '#4F5E53', '#CDDBDE',
              '#382B28', '#98D8C8', '#F7DC6F', '#4F595E', '#85C1E9']

    node_colors = []
    for node in G.nodes():
        comm_id = node_to_community.get(node, 0)
        node_colors.append(colors[comm_id % len(colors)])

    # Tamaños de nodos basados en centralidad
    node_sizes = []
    for node in G.nodes():
        size = degree_centrality[node] * 3000 + 500
        node_sizes.append(size)

    # Grosor de aristas
    edge_widths = []
    for u, v, d in G.edges(data=True):
        width = d['weight'] * 1.0 + 0.5
        edge_widths.append(width)

    # Dibujar la red
    nx.draw(G, pos,
            node_color=node_colors,
            node_size=node_sizes,
            with_labels=True,
            font_size=10,
            font_weight='bold',
            font_color='black',
            edge_color='lightgray',
            width=edge_widths,
            alpha=0.9)

    plt.title("Red de Co-ocurrencias - Layout Circular por Comunidades",
              fontsize=16, fontweight='bold', pad=20)

    # Agregar leyenda
    legend_elements = []
    for i, comm in enumerate(communities):
        color = colors[i % len(colors)]
        label = f"Comunidad {i+1} ({len(comm)} términos)"
        legend_elements.append(plt.Line2D([0], [0], marker='o', color='w',
                                        markerfacecolor=color, markersize=12, label=label))

    plt.legend(handles=legend_elements, loc='center left', bbox_to_anchor=(1, 0.5))
    plt.tight_layout()
    plt.show()

    # ==================== OPCIÓN 2: SOLO CONEXIONES FUERTES ====================
    plt.figure(figsize=(16, 12))

    # Filtrar solo aristas con peso >= 2
    strong_edges = [(u, v) for u, v, d in G.edges(data=True) if d['weight'] >= 2]

    if strong_edges:
        G_strong = G.edge_subgraph(strong_edges)
        pos_strong = nx.spring_layout(G_strong, k=3, iterations=100, seed=42)

        # Colores y tamaños para nodos en grafo filtrado
        strong_node_colors = []
        strong_node_sizes = []

        for node in G_strong.nodes():
            comm_id = node_to_community.get(node, 0)
            strong_node_colors.append(colors[comm_id % len(colors)])
            size = degree_centrality[node] * 4000 + 600
            strong_node_sizes.append(size)

        # Grosor de aristas fuertes
        strong_edge_widths = []
        for u, v, d in G_strong.edges(data=True):
            width = d['weight'] * 1.5 + 0.8
            strong_edge_widths.append(width)

        nx.draw(G_strong, pos_strong,
                node_color=strong_node_colors,
                node_size=strong_node_sizes,
                with_labels=True,
                font_size=11,
                font_weight='bold',
                font_color='black',
                edge_color='gray',
                width=strong_edge_widths,
                alpha=0.9)

        plt.title("Red de Co-ocurrencias - Solo Conexiones Fuertes (peso ≥ 2)",
                  fontsize=16, fontweight='bold')
        plt.tight_layout()
        plt.show()
    else:
        print("No hay conexiones con peso >= 2 para mostrar")

    # ==================== OPCIÓN 3: NODOS MÁS IMPORTANTES ====================
    plt.figure(figsize=(14, 10))

    # Filtrar nodos con grado >= 3
    important_nodes = [node for node, degree in G.degree() if degree >= 3]

    if len(important_nodes) > 1:
        G_important = G.subgraph(important_nodes)
        pos_important = nx.spring_layout(G_important, k=3.5, iterations=100, seed=42)

        # Colores y tamaños para nodos importantes
        important_node_colors = []
        important_node_sizes = []

        for node in G_important.nodes():
            comm_id = node_to_community.get(node, 0)
            important_node_colors.append(colors[comm_id % len(colors)])
            size = degree_centrality[node] * 5000 + 800
            important_node_sizes.append(size)

        # Grosor de aristas
        important_edge_widths = []
        for u, v, d in G_important.edges(data=True):
            width = d['weight'] * 2.0 + 1.0
            important_edge_widths.append(width)

        nx.draw(G_important, pos_important,
                node_color=important_node_colors,
                node_size=important_node_sizes,
                with_labels=True,
                font_size=12,
                font_weight='bold',
                font_color='black',
                edge_color='darkgray',
                width=important_edge_widths,
                alpha=0.9)

        plt.title("Red de Co-ocurrencias - Nodos Más Conectados (grado ≥ 3)",
                  fontsize=16, fontweight='bold')
        plt.tight_layout()
        plt.show()
    else:
        print("No hay suficientes nodos con grado >= 3 para mostrar")

    # ==================== ESTADÍSTICAS BÁSICAS ====================
    print("=" * 60)
    print("ESTADÍSTICAS DE LA RED")
    print("=" * 60)
    print(f"Número de nodos: {G.number_of_nodes()}")
    print(f"Número de enlaces: {G.number_of_edges()}")
    print(f"Densidad: {nx.density(G):.3f}")
    print(f"Número de comunidades: {len(communities)}")
    print(f"Modularidad: {community.modularity(G, communities):.3f}")

    print("\nTop 10 nodos por grado:")
    degree_sorted = sorted(G.degree(), key=lambda x: x[1], reverse=True)
    for node, degree in degree_sorted[:10]:
        print(f"  {node}: {degree}")

    print(f"\nDetalles de comunidades:")
    for i, comm in enumerate(communities, 1):
        print(f"Comunidad {i} ({len(comm)} nodos): {', '.join(list(comm)[:5])}{'...' if len(comm) > 5 else ''}")

    print("\nCoocurrencias más fuertes:")
    edge_weights = [(u, v, d['weight']) for u, v, d in G.edges(data=True)]
    edge_weights.sort(key=lambda x: x[2], reverse=True)
    for u, v, weight in edge_weights[:10]:
        print(f"  {u} - {v}: {weight}")
```
